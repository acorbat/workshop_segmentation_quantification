## Loading 'nuclei (Series 0)'

```python
viewer = napari.Viewer()
```

```python
with BioFile(
    "C:/Users/corba/Documents/workshop_segmentation_quantification/data/nuclei.nd2") as bfile:
    image0_n = bfile.as_array()
viewer.add_image(image0_n, name='nuclei (Series 0)')
viewer.screenshot(path="napari_load.png", canvas_only=False)
```

## median filter

```python
image1_M = nsbatwm.median_filter(image0_n, 3.0)
viewer.add_image(image1_M, name='Result of Median (scipy, nsbatwm)')
viewer.screenshot(path="napari_median.png", canvas_only=False)
```

## subtract background

```python
image1_M.shape
```

```python
image2_R = nsbatwm.subtract_background(image1_M.squeeze(), 40.0)
viewer.add_image(
    image2_R, name='Result of Rolling ball (scikit-image, nsbatwm)')
viewer.screenshot(path="napari_roling_ball.png", canvas_only=False)
```

## threshold otsu

```python
image3_T = nsbatwm.threshold_otsu(image2_R)
viewer.add_labels(
    image3_T, name='Result of Threshold (Otsu et al 1979, scikit-image, nsbatwm)')
viewer.screenshot(path="napari_otsu.png", canvas_only=False)
```

## connected component labeling

```python
image4_C = nsbatwm.connected_component_labeling(image3_T, False)
viewer.add_labels(
    image4_C, name='Result of Connected component labeling (scikit-image, nsbatwm)')
viewer.screenshot(path="napari_label.png", canvas_only=False)
```

## split touching objects

```python
image5_S = nsbatwm.split_touching_objects(image4_C, 2.0)
viewer.add_labels(image5_S, name='Result of Split touching objects (nsbatwm)')
viewer.screenshot(path="napari_split.png", canvas_only=False)
```

## connected component labeling

```python
image6_C = nsbatwm.connected_component_labeling(image5_S, False)
viewer.add_labels(
    image6_C, name='Result of Connected component labeling (scikit-image, nsbatwm) [1]')
viewer.screenshot(path="napari_label_and_split.png", canvas_only=False)
```
